```python
                    # BREADTH FIRST SEARCH (BFS)

from collections import deque

graph = {
    'A' : ['B','C'],
    'B' : ['A','D'],
    'C' : ['A','E'],
    'D' : ['B','E'],
    'E' : ['C','D']
}

def bfs(graph, start):

    visited = set()

    #create an empty queue.
    #BFS follow FIFO: first in,first out,
    queue = deque()

    #here, start = 'A', So A is visited
    visited.add(start)


    #add the starting vertex to the queue.
    #Queue becomes: [A]

    queue.append(start)


    #Continue the loop as long as the queue is not empty.
    while queue:


        #Remove the first vertex from the queue.
        #popleft() removes from the FRONT of the queue.
        vertex = queue.popleft()

        #print the vertex that we just removed.
        #end=" " keeps the output on the same line
        print(vertex, end=" ")


        #Get all the neighbours of current vertex
        #AND PROCESS THEM ONE BY ONE
        for neighbour in graph[vertex]:


            #check whether this neighbour has already been visited.
            #if it is not been visited, enter the if block

            if neighbour not in visited:

                #mark the neighbour as visited
                #This prevents visiting the same vertex again
                visited.add(neighbour)


                #Add the unvisited neighbour to the queue
                # It will be processed later
                queue.append(neighbour)


#Call the BFS function stating from A
bfs(graph, 'A')
```
