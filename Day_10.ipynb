```python
            #HASH TABLE OF LINEAR PROBING

arr = [30, 34, 32, 55, 60]
size = 10
hash_table = [None] * size

for num in arr:
    index = num % size
    hash_table[index] = num 

print(hash_table)
```

```python
            #HASH TABLE WITH COLLISION

arr = [30, 34, 32, 55, 60, 44, 54]
size = 10
hash_table = [None] * size

for ele in arr:

    index = ele % size

    while hash_table[index] is not None:
        index = (index + 1) % size

    hash_table[index] = ele

print(hash_table)
```
