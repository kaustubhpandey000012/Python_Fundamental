```python
def knapsack_algorithm(weights, values, capacity):
    number = len(weights)


    items = []

    for i in range(number):
        ratio = values[i] / weights[i]
        items.append((ratio, weights[i], values[i]))


    items.sort(reverse=True)


    total_values = 0

    for ratio, weight, value in items:
        if capacity == 0:
            break

        if weight <= capacity:
            capacity -= weight
            total_values += value


            print(f"Take full item: Weight={weight}," f"value={value}")

        else:
            fraction = capacity / weight
            gained_value = value * fraction

            total_values += gained_value

            print(f"Take {fraction:.2%}of item;"
                  f"Weight={capacity},"
                  f"value={gained_value:.2f}"
                  )


            capacity = 0
    return total_values


weights = [10, 20, 30]
values = [60, 100, 120]
capacity = 50

answer =  knapsack_algorithm(weights, values, capacity)


print("Maximum values:", answer)
 
```
