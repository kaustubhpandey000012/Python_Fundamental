```python
     #Palindrome by using Slicing
string = input("Enter a string: ")

if string == string[::-1]:
    print("its palindrome")

else:
    print("it is not a palindrome")
```

```python
    #Palindrone by two pointer Method
string = input("Enter a string: ")

left = 0
right = len(string) - 1

palindrome = True

while left < right:

    if string[left] != string[right]:
        palindrome = False
        break

    left +=1
    right -=1

if palindrome:
    print("It's a palindrome")
else:
    print("It is not a palindrome")
```
