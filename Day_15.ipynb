```python
class Solution(object):
    def maxArea(self, height):
        """
        :type height: List[int]
        :rtype: int
        """

        max_water = 0
        l = len(height)

        for i in range(l):
            for j in range(i+1, l):
                water = min(height[i], height[j]) * (j-i)
                max_water =  max(max_water, water)

        return max_water

solution = Solution()
height = [1,8,6,2,5,4,8,3,7]
print(solution.maxArea(height))
```
