```python
         # Class car
class car:

    def __init__(self,brand, color, top_speed):
        self.brand = brand
        self.color = color
        self.top_speed = top_speed
        self.current_speed = 0

    def accelarate(self):
        self.current_speed += 120
        print(f"The {self.color} {self.brand} is now moving at {self.current_speed} km/h")


car1 = car("Ferrari","Red",400)
car2 = car("BMW","Green",200)


car1.accelarate()
car2.accelarate()
```

```python
class Animal:
    def __init__(self, name, color, age):
        self.name = name
        self.color = color
        self.age = age

    def bark(self):
        self.age += 2
        print(f"The {self.color} {self.name} is now {self.age} years old.")

Animal1 = Animal("Dog","white",5)
Animal2 = Animal("wolf","silver",12)
Animal3 = Animal("leopard","brown",16)

Animal1.bark()
Animal2.bark()
Animal3.bark()
```
